```python
import turtle
import math

# 1. Настройка экрана
screen = turtle.Screen()
screen.setup(600, 600)
# Устанавливаем систему координат: x от -10 до 10, y от -10 до 100
screen.setworldcoordinates(-10, -10, 10, 100)

# 2. Настройка черепашки
t = turtle.Turtle()
t.speed(0) # Максимальная скорость
t.hideturtle() # Скрыть черепашку
t.penup()

# 3. Рисование осей (опционально)
t.goto(-10, 0)
t.pendown()
t.goto(10, 0)
t.penup()
t.goto(0, -10)
t.pendown()
t.goto(0, 100)
t.penup()

# 4. Построение графика
t.color("blue")
t.pensize(2)

for x in range(-10, 11):
    y = x**3
    if x == -10:
        t.goto(x, y)
        t.pendown()
    else:
        t.goto(x, y)

# Завершение
turtle.done()
```

```python
a = "abcdefjhijklmnopqrstuvwxyz"

print(list(enumerate(a)))
diid = lambda s: [i for i in s]
print(diid(a))


def length(l):
    summ = 0
    for i in l:
        summ += 1
    return summ


def enmrt(l):
    symbols = diid(l)
    lelen = length(symbols)
    numbers = [i for i in range(lelen)]
    result = []
    for i in range(lelen):
        result.append((symbols[i], numbers[i]))
    return result


print(length(a))
print(enmrt(a))
```
